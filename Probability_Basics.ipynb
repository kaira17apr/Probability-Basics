**Q1. A die is rolled. What is the probability of getting:**

**(a) An even number**

**(b) A number greater than 4**

**Ans-**

A fair die has 6 faces: {1, 2, 3, 4, 5, 6}.

Each outcome is equally likely, so the probability of any single outcome is
1/6.

**(a) Probability of getting an even number**


*   Even numbers on a die: {2, 4, 6}

*   Number of favorable outcomes = 3
*   Total possible outcomes = 6

𝑃(even number)= 3/6= 1/2

**(b) Probability of getting a number greater than 4**



*   Numbers greater than 4: {5, 6}

*   Number of favorable outcomes = 2
*   Total possible outcomes = 6

𝑃(number > 4)= 2/6

𝑃 = 1/3







**Q2. In a class of 50 students:**

*   **20 like Mathematics (M)**
*   **15 like Science (S)**
*   **5 like both subjects**

**What is the probability that a student chosen at random likes Mathematics or Science?**

**Ans-**

**Step 1: Define the sets**



*   Total students = 50
*   Students who like Mathematics (M) = 20
*   Students who like Science (S) = 15
*   Students who like both (M ∩ S) = 5

**Step 2: Apply the union formula**

The number of students who like Mathematics or Science is given by:

𝑛(𝑀∪𝑆)=𝑛(𝑀)+𝑛(𝑆)−𝑛(𝑀∩𝑆)

𝑛(𝑀∪𝑆)=20+15−5=30

So, 30 students like Mathematics or Science.

**Step 3: Probability**

Probability = Favorable outcomes/Total outcomes

𝑃(M or S)= 30/50

𝑃= 3/5




**Q3. A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability that the next ball is also red (without replacement)?**

**Ans-**

**Step 1: Initial setup**



*   Bag contains 3 red and 2 blue balls → total = 5 balls.

*   First ball drawn is red (without replacement).

*   After removing one red ball, the bag now has:
      *   Red balls = 2
      *   Blue balls = 2
      *   Total balls left = 4

**Step 2: Probability of second ball being red**

Now we want the probability that the next ball is also red:

𝑃(second red | first red)= Remaining red balls / Remaining total balls

𝑃= 2/4

𝑃= 1/2

**Q4. The population of a school is divided into 60% boys and 40% girls. If you want equal representation of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling? Why?**

**Ans-**

I use **Stratified Sampling** rather than **Simple Random Sampling**.

**Explanation**



*   **Simple Random Sampling (SRS):** Every student has an equal chance of being selected. Since the population is 60% boys and 40% girls, an SRS sample will likely reflect this same proportion. That means boys will be overrepresented compared to girls.
*   **Stratified Sampling:** The population is divided into strata (here, boys and girls). Then you select equal numbers (or proportional numbers, depending on the goal) from each stratum. If you want equal representation, you deliberately choose the same number of boys and girls in the sample.




**Q5. The average height of 1000 students = 160 cm. A sample of 100 students shows an average height = 158 cm.**

**Find the sampling error.**

**Ans-**

**Step 1: Recall the formula**

Sampling error is the difference between the sample mean and the population mean:

Sampling Error =𝑥ˉ−𝜇

Where:


*   𝑥ˉ= sample mean
*   𝜇 = population mean

**Step 2: Substitute values**



*   Population mean (𝜇) = 160 cm
*   Sample mean (𝑥ˉ) = 158 cm

Sampling Error =158−160=−2

**Step 3: Interpret**



*   The sampling error is **–2 cm**.
*   This means the sample average height is **2 cm lower** than the population average.

**Q7.  In a group of 100 students:**

*   **40 like Cricket (C)**
*   **30 like Football (F)**
*  **10 like both Cricket and Football**

**Find the probability that a student likes at least one sport.**

**Ans-**

**Step 1: Define the sets**



*   Total students = 100
*   Students who like Cricket (C) = 40
*   Students who like Football (F) = 30
*   Students who like both (C ∩ F) = 10

**Step 2: Apply the union formula**

The number of students who like at least one sport (Cricket or Football) is:

𝑛(𝐶∪𝐹)=𝑛(𝐶)+𝑛(𝐹)−𝑛(𝐶∩𝐹)

𝑛(𝐶∪𝐹)=40+30−10=60

So, 60 students like at least one sport.

**Step 3: Probability**

𝑃(at least one sport)= 𝑛(𝐶∪𝐹)/Total students

𝑃=60/100

𝑃=3/5

**Q8. From a deck of 52 cards, two cards are drawn without replacement. What is the probability that both are Aces?**

**Ans-**

**Step 1: Total cards and Aces**


*   A standard deck has 52 cards.
*   Number of Aces = 4.

**Step 2: Probability of first card being an Ace**

𝑃(first Ace)=4/52=1/13

**Step 3: Probability of second card being an Ace (without replacement)**

After one Ace is drawn,



*   Remaining Aces = 3
*   Remaining cards = 51

𝑃(second Ace | first Ace)=3/51

**Step 4: Joint probability**

Since both events must happen:

𝑃(both Aces)=4/52×3/51

𝑃=12/2652=1/221




**Q9. A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the probability that all are non-defective?**

**Ans-**

*   Probability of a bulb being non-defective = 1−0.02=0.98

We want the probability that all 5 bulbs are non-defective.

Since each draw is independent:

𝑃(all non-defective)=(0.98)⁵


**Step 3: Calculation**

(0.98)⁵=0.9039 (approx)

**Q10. Differentiate between discrete and continuous random variables with examples.**

**Ans-**

**Discrete Random Variables**



*   **Definition:** Variables that take on countable values (finite or countably infinite).

*   **Nature:** Outcomes are distinct and separate; you can list them individually.

*   **Examples:**
    *   Number of students absent in a class (0, 1, 2, …).


    *   Number of heads in 10 coin tosses.


    *   Roll of a die (1, 2, 3, 4, 5, 6).


**Continuous Random Variables**



*   **Definition:** Variables that take on uncountably infinite values within an interval.

*   **Nature:** Outcomes are measured, not counted; values can take any real number in a range.

*   **Examples:**
    *   Height of students (e.g., 160.2 cm, 160.25 cm, etc.).


    *   Time taken to run a race (e.g., 12.31 seconds, 12.315 seconds).


    *   Temperature in a city on a given day.
